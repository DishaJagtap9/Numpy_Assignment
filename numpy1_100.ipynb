````python
##### 88. How to implement the Game of Life using numpy arrays? (★★★)
#`No hints provided...
import numpy as np
def game_of_life(Z):

    # Count neighbors
    N = (
        np.roll(Z, 1, 0) + np.roll(Z, -1, 0) +
        np.roll(Z, 1, 1) + np.roll(Z, -1, 1) +
        np.roll(np.roll(Z, 1, 0), 1, 1) +
        np.roll(np.roll(Z, 1, 0), -1, 1) +
        np.roll(np.roll(Z, -1, 0), 1, 1) +
        np.roll(np.roll(Z, -1, 0), -1, 1)
    )

    # Apply Game of Life rules
    return ((N == 3) | ((Z == 1) & (N == 2))).astype(int)

# Initial grid
Z = np.array([
    [0,1,0],
    [0,1,0],
    [0,1,0]
])

print("Original:")
print(Z)

Z = game_of_life(Z)

print("\nNext Generation:")
print(Z)
````

````python
##### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?
#`hint: np.cumsum, np.interp`
import numpy as np

# Path coordinates
X = np.array([0, 1, 2, 4, 7])
Y = np.array([0, 2, 1, 3, 2])

# Distance between consecutive points
d = np.sqrt(np.diff(X)**2 + np.diff(Y)**2)

# Cumulative distance along the path
s = np.insert(np.cumsum(d), 0, 0)

# Number of equidistant samples
n = 10

# Equally spaced distances
s_new = np.linspace(0, s[-1], n)

# Interpolate X and Y coordinates
X_new = np.interp(s_new, s, X)
Y_new = np.interp(s_new, s, Y)

print("X_new =", X_new)
print("Y_new =", Y_new)
````
