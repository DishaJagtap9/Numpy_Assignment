## Assignment 5 — Logical Operations & Filtering

````python
#1. Find all days where temperature was **above 35°C** (heat wave).
import numpy as np
temps = np.array([22, 35, 28, 41, 19, 37, 30, 25, 33, 18,
                  40, 29, 23, 36, 31, 27, 38, 20, 24, 34,
                  21, 39, 26, 32, 17, 43, 28, 35, 22, 30])
print(temps[temps>35])
````

````python
#2. Find all days where temperature was **below 20°C** (cold day).
print(temps[temps<20])
````

````python
#3. Find days where temperature was **between 25°C and 35°C** (comfortable).
print(temps[(temps >= 25) & (temps <= 35)])
````

````python
#4. Count how many days were comfortable.
count = np.sum((temps >= 25) & (temps <= 35))

print(count)
````

````python
#5. Find days that were either **heat waves** or **cold days** (extreme days).
extreme_days = temps[(temps > 35) | (temps < 20)]

print(extreme_days)
````

````python
#6. Were **all** days above 15°C? Use `np.all()`.

result = np.all(temps > 15)

print(result)
````

````python
#7. Were **any** days above 42°C? Use `np.any()`.
result=np.any(temps>42)
print(result)
````

## Assignment 6 — NaN & Inf Handling (Data Cleaning)

Given sensor data from an IoT device:
```python
sensor = np.array([23.1, np.nan, 24.5, np.inf, 22.8, np.nan,
                   25.3, -np.inf, 21.9, 24.0, np.nan, 23.7])
```

1. Count the number of `NaN` values.
2. Count the number of `Inf` values (both +inf and -inf).
3. Extract only the valid (finite) readings.
4. Replace NaN values with the **mean** of all finite values.
5. Replace Inf values with the **max** finite value and -Inf with the **min** finite value.
6. Calculate `nanmean`, `nanmin`, `nanmax`, and `nanstd` of the sensor array.
7. **Bonus**: Write a single function `clean_sensor_data(arr)` that handles all of the above and returns a clean array.

````python
#1. Count the number of `NaN` values.
sensor = np.array([23.1, np.nan, 24.5, np.inf, 22.8, np.nan,
                   25.3, -np.inf, 21.9, 24.0, np.nan, 23.7])
count=np.sum(np.isnan(sensor))
print("number of NAN values",count)
````

````python
#2. Count the number of `Inf` values (both +inf and -inf).
count=np.sum(np.isinf(sensor))
print(count)
````

````python
#3. Extract only the valid (finite) readings.
valid_finite=sensor[np.isfinite(sensor)]
print(valid_finite)
````

````python
#4. Replace NaN values with the **mean** of all finite values.
mean_value =np.mean(sensor[np.isfinite(sensor)])
sensor[np.isnan(sensor)]=mean_value
print(sensor)
````

````python
#5. Replace Inf values with the **max** finite value and -Inf with the **min** finite value.
finite_values=sensor[np.isfinite(sensor)]
max_finite=np.max(finite_values)
min_finite=np.min(finite_values)
sensor[np.isposinf(sensor)]=max_finite
sensor[np.isneginf(sensor)]=min_finite
print(sensor)
````

````python
#6. Calculate `nanmean`, `nanmin`, `nanmax`, and `nanstd` of the sensor array.
import numpy as np

mean_value = np.nanmean(sensor)
min_value = np.nanmin(sensor)
max_value = np.nanmax(sensor)
std_value = np.nanstd(sensor)

print("nanmean:", mean_value)
print("nanmin :", min_value)
print("nanmax :", max_value)
print("nanstd :", std_value)
````

````python
#7. **Bonus**: Write a single function `clean_sensor_data(arr)` that handles all of the above and returns a clean array.
def clean_sensor_data(arr):
    arr=arr.copy()
    finite_values=arr[np.isfinite(arr)]

    max_finite=np.max(finite_values)
    min_finite=np.min(finite_values)
    mean_finite=np.mean(finite_values)
    arr[np.isposinf(arr)]=max_finite
    arr[np.isneginf(arr)]=min_finite
    
    arr[np.isnan(arr)]=mean_finite

    return arr
cleaned_sensor = clean_sensor_data(sensor)

print("Original Sensor Data:")
print(sensor)

print("\nCleaned Sensor Data:")
print(cleaned_sensor)
````

## 🏆 Capstone Challenge

You receive a dataset of monthly sales figures for 6 products across 12 months. Some values are missing (NaN) and some are erroneous (Inf or negative):

```python
sales = np.array([
    [1500, 1800, np.nan, 2100, 1950, 2300, np.nan, 1700, 2000, 2150, 1600, 1900],
    [800,  np.nan, 950, 1100, 850,  np.inf, 900,  1050, np.nan, 800,  920,  1000],
    [3000, 3200, 2900, np.nan, 3100, 3500, 3300, np.nan, 3050, 3400, 3150, 3600],
    [450,  500,  np.nan, 600,  -100, 550,  np.nan, 480,  520,  -np.inf, 490, 530],
    [2200, 2100, 2400, 2300, np.nan, 2500, 2350, 2450, 2100, 2300, np.nan, 2600],
    [100,  120,  90,   np.nan, 130, 110,  140,   np.nan, 100, 120,  np.inf, 115]
])
```

**Tasks:**
1. Print the shape of the dataset.
2. Count total NaN, Inf, and negative values.
3. Clean the data: replace NaN with row-wise nanmean, replace Inf/-Inf/negatives with 0.
4. After cleaning, find the product (row) with the highest total annual sales.
5. Find which month (column index) had the highest total sales across all products.
6. Find which months had total sales above the overall mean.
7. Cast the cleaned array to `float32` to save memory. How much memory was saved?

````python
#1. Print the shape of the dataset.
sales = np.array([
    [1500, 1800, np.nan, 2100, 1950, 2300, np.nan, 1700, 2000, 2150, 1600, 1900],
    [800,  np.nan, 950, 1100, 850,  np.inf, 900,  1050, np.nan, 800,  920,  1000],
    [3000, 3200, 2900, np.nan, 3100, 3500, 3300, np.nan, 3050, 3400, 3150, 3600],
    [450,  500,  np.nan, 600,  -100, 550,  np.nan, 480,  520,  -np.inf, 490, 530],
    [2200, 2100, 2400, 2300, np.nan, 2500, 2350, 2450, 2100, 2300, np.nan, 2600],
    [100,  120,  90,   np.nan, 130, 110,  140,   np.nan, 100, 120,  np.inf, 115]
])
print(sales.shape)
````

````python
#2. Count total NaN, Inf, and negative values.
nan_count=np.sum(np.isnan(sales))
Inf_count=np.sum(np.isinf(sales))
negative_count = np.sum(sales < 0)
print("count of NaN:",nan_count)
print("count of inf",Inf_count)
print("count of nagative ",negative_count)
````

````python
#3. Clean the data: replace NaN with row-wise nanmean, replace Inf/-Inf/negatives with 0.
import numpy as np
row_means=np.nanmean(sales,axis=1)
for i in range (sales.shape[0]):
    sales[i,np.isnan(sales[i])]=row_means[i]
sales[np.isinf(sales)]=0
sales[sales<0]=0
print(sales)
````

````python
#4. After cleaning, find the product (row) with the highest total annual sales.
row_totals = np.sum(sales, axis=1)


highest_product = np.argmax(row_totals)

print("Total sales of each product:", row_totals)
print("Product with highest total annual sales:", highest_product)
print("Highest total annual sales:", row_totals[highest_product])
````

````python
#5. Find which month (column index) had the highest total sales across all products.
column_totals=np.sum(sales,axis=0)
highest_month=np.argmax(column_totals)
print("total sales for each month:",column_totals)
print("month (column index)with highest total sales:",highest_product)
print("highest total sales",column_totals[highest_month])
````

````python
#6. Find which months had total sales above the overall mean.
column_totals=np.sum(sales,axis=0)
Overall_mean=np.mean(column_totals)
months_above_mean= np.where(column_totals>Overall_mean)[0]
print("monthly total:",column_totals)
print("overall mean:",Overall_mean)
print("month above overall mean:",months_above_mean)
````

````python
#7. Cast the cleaned array to `float32` to save memory. How much memory was saved?
memory_before=sales.nbytes
sales_float32=sales.astype(np.float32)
memory_after=sales_float32.nbytes
memory_saved=memory_before-memory_after

print("memory before:",memory_before,"bytes")
print("memory after:",memory_after,"bytes")
print("memory saved:",memory_saved,"bytes")
````
