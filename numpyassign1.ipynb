````python
#**Q2.** Generate a `(50, 4)` dataset where:  
#- Features are random normal data  
#- True relationship: `y = 1.5*x1 - 2.0*x2 + 0.5*x3 + 3.0*x4 + noise`  

#Estimate weights using the Normal Equation with a bias term. Report MSE.
rng=np.random.default_rng(7)
#genrate the feature
X=rng.normal(size=(50,4))
#Genrate noise
noise=rng.normal(scale=0.1,size=50)
#true target
y = (
    1.5 * X[:, 0]
    - 2.0 * X[:, 1]
    + 0.5 * X[:, 2]
    + 3.0 * X[:, 3]
    + noise
)
#add bias column
X_bias =np.c_[np.ones(X.shape[0]),X]
#Noraml Equation
w=np.linalg.inv(X_bias.T@ X_bias)@ X_bias.T@y
print("Estimated weight:")
print(w)
#prediction 
y_pred=X_bias @ w
#MSE
mse=np.mean((y-y_pred)**2)
print("MSE:",mse)


````

````python
#*Q3.** Compute cosine similarity between all pairs of these three word vectors (simplified word embeddings):
#```python
#king   = np.array([0.8, 0.3, 0.5])
#queen  = np.array([0.7, 0.4, 0.6])
#apple  = np.array([0.1, 0.9, 0.2])

#Which pair is most similar? Does this match your intuition?
import numpy as np
king   = np.array([0.8, 0.3, 0.5])
queen  = np.array([0.7, 0.4, 0.6])
apple  = np.array([0.1, 0.9, 0.2])
def consine_similarity(a,b):
    return np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))

# pairwise similarities
king_queen=consine_similarity(king,queen)
king_apple=consine_similarity(king,apple)
queen_apple=consine_similarity(queen,apple)

print("king queen:",king_queen)
print("king_apple:",king_apple)
print("queen-apple:",queen_apple)

````
